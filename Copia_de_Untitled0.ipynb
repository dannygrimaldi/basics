```python
import msoffcrypto
import io
import pandas as pd
from typing import Optional

def excel_protegido_a_dataframe(
    ruta_archivo: str,
    contraseña: str,
    hoja: Optional[str] = None,
) -> pd.DataFrame:
    """
    Convierte un archivo Excel protegido por contraseña en un DataFrame de Pandas.

    Parámetros:
    -----------
    ruta_archivo : str
        Ruta completa del archivo Excel protegido.
    contraseña : str
        Contraseña para desbloquear el archivo.
    hoja : str, opcional
        Nombre de la hoja a cargar. Si es None, carga la primera hoja.
    Retorna:
    --------
    pd.DataFrame
        DataFrame con los datos del Excel.

    Excepciones:
    ------------
    FileNotFoundError
        Si el archivo no existe.
    msoffcrypto.exceptions.InvalidKeyError
        Si la contraseña es incorrecta.
    Exception
        Para otros errores durante el procesamiento.
    """
    try:
        # Abrir y desencriptar el archivo
        with open(ruta_archivo, 'rb') as f:
            office_file = msoffcrypto.OfficeFile(f)
            office_file.load_key(password=contraseña)
            decrypted_data = io.BytesIO()
            office_file.decrypt(decrypted_data)

            # Reiniciar el puntero del BytesIO
            decrypted_data.seek(0)

            # Leer el archivo Excel
            if hoja:
                df = pd.read_excel(decrypted_data, sheet_name=hoja)
            else:
                df = pd.read_excel(decrypted_data)

            return df

    except FileNotFoundError:
        raise FileNotFoundError(f"Archivo no encontrado: {ruta_archivo}")
    except msoffcrypto.exceptions.InvalidKeyError:
        raise ValueError("Contraseña incorrecta o archivo no protegido")
    except Exception as e:
        raise Exception(f"Error al procesar el archivo: {str(e)}")


# Ejemplo de uso
if __name__ == "__main__":
    try:
        # Uso básico
        df = excel_protegido_a_dataframe(
            ruta_archivo='/content/sample_data/Calculos_Proyecto.xlsx',
            contraseña='Grogu05'
        )
        print("DataFrame cargado exitosamente:")
        print(df.head())

        # Uso avanzado especificando hoja
        df_hoja2 = excel_protegido_a_dataframe(
            ruta_archivo='/content/sample_data/Calculos_Proyecto.xlsx',
            contraseña='Grogu05',
            hoja='Hoja1'
        )
        print(df.head())

    except Exception as e:
        print(f"Error: {e}")
```
